```python
# 1. Days to seconds
days = float(input("Enter the number of days: ")) # Ask the user to input a number of days, convert input string to a float

# Convert days to seconds:
# 1 day = 24 hours, 1 hour = 60 minutes, 1 minute = 60 seconds
seconds = days * 24 *60 * 60

# Display the result to the user
print(f"{days} days is equal to {seconds} seconds.")
```

```python
# 2. Volume of a sphere
import math # Import the math module to use the pi constant

radius = float(input("Enter the radius: ")) # Ask the user to input the radius of the sphere, convert input string to a float

# Calculate the volume using the formula: 4/3 * pi * radius^3
volume = (4/3) * math.pi * radius**3 # The ** operator is used for exponentiation

# Display the result to the user
print(f"The volume of the sphere is {volume:.2f} cubic units.")
```

```python
# 3. Area and Perimeter of a square
# Calculate the area of a square given the side length
def area(side):
    return side ** 2 # area = side * side

# Function to calculate perimeter of a sqaure given the side length
def perimeter(side):
    return 4 * side # perimeter = 4 * side length

# Ask the user to input the side length of the square
side_length = float(input("Enter the side length of the square: "))

# Call the area() function and display the result
print(f"Area: {area(side_length)}")

# Call the perimeter() function and display the result
print(f"Perimeter: {perimeter(side_length)}")
```

```python
# 4. Uppercase or Lowercase check
# Function that takes a character and returns whether it's uppercase, lowercase, or not a letter
def check_case(char):
    if char.isupper():        # Check if the character is an uppercase letter
        return "Uppercase"
    elif char.islower():      # Check if the character is a lowercase letter
        return "Lowercase"
    else:                     # Handles numbers, symbols, or non-letter input
        return "Not a letter"

# Ask the user to input a character
character = input("Enter a character: ")

# Call the check_case() function and display the result
print(check_case(character))
```

```python
#5. Pseudocode conversion
# Initialize variables
x = 0
y = 20

# Python doesn't have a built-in "repeat...until" loop,
# so we simulate it using while True with a break condition at the end
while True:
    y = y - 4          # Subtract 4 from y
    x = x + 2 / y       # Add 2/y to x
    
    if y < 6:           # Check the exit condition (UNTIL y IS LESS THAN 6)
        break            # Exit the loop once the condition is true

# Display the final value of x
print(x)
```

```python
# 6. Average of 5 values
values = []  # Create an empty list (array) to store the input values

# Loop 5 times to collect 5 values from the user
for i in range(5):
    num = float(input(f"Enter value {i + 1}: "))  # Ask for and convert input to float
    values.append(num)  # Add the entered value to the list

# Calculate the average: sum of all values divided by the count of values
average = sum(values) / len(values)

# Display the calculated average
print(f"The average of the values is: {average}")
```

```python
# Codility Task 
def solution(A):
    n = len(A)
    left = 0 # Initialize left pointer to the start of the array
    right = n - 1 # Initialize right pointer to the end of the array

    distinct_count = 0
    last_abs = -1 # Track the last absolute value to avoid counting duplicates
    
    # Loop until the left pointer is greater than the right pointer
    while left <= right:
        abs_left = abs(A[left]) # Get the absolute value of the left element
        abs_right = abs(A[right]) # Get the absolute value of the right element
        
        # Take the larger of the two absolute values
        if abs_left > abs_right:
            current_abs = abs_left 
            left += 1 
        else:
            current_abs = abs_right
            right -= 1
        
        # Only count it if it's different from the last counted absolute value
        if current_abs != last_abs:
            distinct_count += 1
            last_abs = current_abs

    return distinct_count
```
